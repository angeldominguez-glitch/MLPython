### Repaso de Python para datos y la potencia de `NumPy`

Al trabajar con datos, debemos ser eficientes. Mientras que las listas de Python son versátiles, la biblioteca `NumPy` introduce los «arrays», que permiten representar colecciones de datos del mismo tipo en varias dimensiones. Un hito técnico clave es que el procesamiento de estos «arrays» es hasta 50 veces más rápido que las listas tradicionales, algo vital para el cálculo numérico y el análisis de grandes volúmenes de información.

Para comprobar por qué esto es tan relevante, vamos a plantear una situación real: calcular la temperatura media registrada durante una semana en Valladolid. Primero implementaremos la lógica pura utilizando solo las estructuras básicas de Python y, a continuación, veremos cómo `NumPy` simplifica drásticamente el proceso.

---

```python
# PRÁCTICA 1: Cálculo de la media SIN librerías externas (Python Puro)

## Simulamos los registros térmicos diarios (en grados Celsius) en Valladolid

temperaturas_valladolid_lista = [12.5, 14.0, 11.2, 9.5, 15.1, 16.3, 13.8]

## Necesitamos inicializar un contador y recorrer iterativamente la lista

suma_temperaturas = 0
for temp in temperaturas_valladolid_lista:
    suma_temperaturas += temp

## Finalmente, dividimos entre el número total de observaciones

media_temperaturas_nativa = suma_temperaturas / len(temperaturas_valladolid_lista)
print(f"La temperatura media en Valladolid es: {media_temperaturas_nativa:.2f} °C")
```

---

```python
# PRÁCTICA 2: El mismo cálculo utilizando NumPy

import numpy as np

## Convertimos nuestra lista tradicional en un 'array' de NumPy optimizado

temperaturas_valladolid_array = np.array(temperaturas_valladolid_lista)

## Aprovechamos el método 'mean()', escrito internamente en lenguaje C para máxima velocidad

media_temperaturas_numpy = temperaturas_valladolid_array.mean()
print(f"La temperatura media en Valladolid es: {media_temperaturas_numpy:.2f} °C")
```

---

Como habéis podido observar, `NumPy` nos exime de escribir algoritmos de bajo nivel como bucles sumatorios, permitiéndonos delegar la carga computacional en herramientas hiperoptimizadas.

## 3. Sesión 2: Domando los datos con `Pandas` (3 horas)

`Pandas` es, sin duda, la columna vertebral de la ciencia de datos en Python. Su función estratégica es transformar archivos planos (con extensión `.csv`) en estructuras de conocimiento llamadas «_dataframes_». Para un desarrollador de IA, un _dataframe_ es la representación en Python de lo que denominamos técnicamente «tablas de datos», permitiendo una manipulación que las hojas de cálculo tradicionales no pueden alcanzar.

Frente a una hoja de cálculo, `Pandas` ofrece escalabilidad y automatización. Podéis procesar millones de registros sin que el sistema se bloquee —algo común en Excel— y definir procesos de limpieza que se ejecutan sobre miles de archivos con un solo clic. Es la diferencia entre el trabajo manual y la ingeniería de datos.

A modo de ejemplo, imaginemos que el Ayuntamiento nos proporciona un archivo CSV con el estado del servicio de alquiler de bicicletas (BIKI). Veamos primero el esfuerzo que supondría procesar, limpiar y filtrar este archivo usando exclusivamente la librería nativa de manipulación de archivos.

---

```python
# PRÁCTICA 3: Análisis de archivos planos SIN Pandas (Python Puro)
import csv
import io

## Simulamos la lectura de un CSV extraído de un portal de Datos Abiertos
csv_biki_valladolid = """ID_Estacion,Nombre_Parada,Barrio,Bicis_Disponibles
1,Plaza Mayor,Centro,15
2,Estación Campo Grande,Delicias,2
3,Feria de Muestras,Huerta del Rey,12
4,Hospital Clínico,Rondilla,1
5,Marcos Fernández,Parquesol,0"""

print("--- Procesamiento manual de alertas de baja disponibilidad ---")

## Debemos gestionar la apertura de memoria y la lectura de diccionarios a mano

with io.StringIO(csv_biki_valladolid) as archivo:
    lector_csv = csv.DictReader(archivo)
    for fila in lector_csv:
        # Los datos se leen como texto, por lo que es obligatorio hacer un casting (int)
        bicis_actuales = int(fila['Bicis_Disponibles'])
        if bicis_actuales < 5:
            print(f"ALERTA BIKI: Quedan solo {bicis_actuales} bicis en {fila['Nombre_Parada']} ({fila['Barrio']}).")
```

---

Aunque funcional, este enfoque es propenso a errores en cuanto el volumen o la complejidad del filtrado aumenta. Con `Pandas`, esta operación se reduce a sentencias vectorizadas mucho más limpias y potentes.

Entre el arsenal de métodos que usaremos habitualmente se encuentran `head()`, que nos ofrece una primera toma de contacto visualizando las filas iniciales; `info()`, esencial para comprobar la integridad técnica al mostrar tipos de datos y detectar valores nulos; y `describe()`, que genera de inmediato un perfil estadístico calculando medias, máximos y mínimos.  

---

```python
# PRÁCTICA 4: Carga y análisis automatizado CON Pandas
import pandas as pd

## Carga de un dataset de ejemplo
## Pandas asimila el texto directamente y deduce
## (infiere) los tipos de datos numéricos y de texto

df_biki = pd.read_csv(io.StringIO(csv_biki_valladolid))
print("1. Inspección inicial con el método head():")

## head() es ideal para la primera toma de contacto
## visualizando filas iniciales

display(df_biki.head())
print("\n2. Filtrado avanzado de información:")

## Podemos realizar consultas complejas con una sintaxis limpia
## Filtramos directamente devolviendo un subconjunto
## ('dataframe') sin usar bucles

paradas_baja_disponibilidad = df_biki[df_biki['Bicis_Disponibles'] < 5]
display(paradas_baja_disponibilidad[['Nombre_Parada', 'Bicis_Disponibles']])
```

## 4. Sesión 3: El arte de la visualización exploratoria (3 horas)

La visualización es la herramienta de descubrimiento por excelencia; es la que nos permite comunicar patrones ocultos de forma efectiva. En el ecosistema Python, contamos con dos pilares: `Matplotlib`, especializada en la creación y personalización exhaustiva de gráficos en dos dimensiones, y `Seaborn`, que actúa como una interfaz de alto nivel que simplifica la creación de gráficos estadísticos informativos y visualmente atractivos.

### Gráficos fundamentales para el análisis

* __Gráficos de barras__: La opción predilecta para comparar categorías.
* __Gráficos de dispersión__: Esenciales para encontrar relaciones entre variables y detectar valores atípicos que podrían arruinar un modelo de IA.

La visualización es el puente de comunicación entre el dato y la decisión. Una elección incorrecta del gráfico —como usar una línea para categorías independientes— puede inducir a conclusiones erróneas. `Seaborn` nos protege de estos fallos al gestionar la complejidad estadística de fondo, permitiendo que el desarrollador se enfoque en interpretar la _historia_ que cuentan los datos.

Para aplicar estos conceptos, vamos a plantear una situación cercana: analizar el uso del transporte público. Visualizaremos mediante un diagrama de barras la densidad de uso mensual de AUVASA (Autobuses Urbanos de Valladolid), discriminando por el tipo de línea contratada por los ciudadanos.

---

```python
# PRÁCTICA 5: Descubriendo patrones visuales con Seaborn

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

## 1. Preparación del dataset: Simulamos datos abiertos de movilidad urbana de Valladolid

datos_movilidad_auvasa = {
    'Linea_Bus': ['Ordinarias (1,2,3...)', 'Ordinarias (1,2,3...)', 'Búho (B1, B2...)', 'Búho (B1, B2...)'],
    'Pasajeros_Mensuales': [450000, 480000, 35000, 41000],
    'Mes_Registro': ['Noviembre', 'Diciembre', 'Noviembre', 'Diciembre']
}
df_auvasa = pd.DataFrame(datos_movilidad_auvasa)

## 2. Configuración estética de la gráfica utilizando Seaborn

sns.set_theme(style="whitegrid") # Establecemos un fondo claro cuadriculado para facilitar la lectura

## 3. Creación del gráfico

plt.figure(figsize=(9, 5)) # Dimensiones del lienzo

## Generamos un gráfico de barras comparando categorías (Líneas) segmentadas por otra variable (Mes)

grafico_barras = sns.barplot(
    x='Linea_Bus',
    y='Pasajeros_Mensuales',
    hue='Mes_Registro',
    data=df_auvasa,
    palette="muted"
)

## 4. Personalización y etiquetado

plt.title('Densidad mensual de pasajeros en AUVASA por tipología de servicio', fontsize=14, pad=15)
plt.xlabel('Tipología de Servicio (AUVASA)', fontsize=12)
plt.ylabel('Volumen de Pasajeros', fontsize=12)

## Despliegue de la visualización

plt.show()
```
