### 1. Tipos dinámicos
Python es de [tipado dinámico](https://es.wikipedia.org/wiki/Tipado_din%C3%A1mico), lo que significa que una variable puede almacenar datos de cualquier tipo, y ese tipo puede cambiar .

Por ejemplo, si creo una variable numérica, posteriormente puedo cambiar esa variable a una cadena de texto:

```python
suscriptores_activos = 45200
suscriptores_activos = "Cuarenta y cinco mil"
```

Que algo se pueda hacer no tiene por qué ser, necesariamente, una buena idea. En casi todas las circunstancias es una __mala idea__ cambiar el tipo de una variable. Lo _correcto_ suele ser, simplemente, crear una nueva variable. Por ejemplo:

```python
suscriptores_activos = 45200
suscriptores_descripcion = "Cuarenta y cinco mil"
```

Los lenguajes que no son de tipado dinámico son de tipado estático, como Go y TypeScript. En un lenguaje de tipado estático, si intentas asignar un valor a una variable de tipo incorrecto, obtendrás un error de compilación y el programa no se ejecutará.

Si Python fuera de tipado estático, el primer ejemplo anterior no permitiría la segunda línea, `suscriptores_activos = "Cuarenta y cinco mil"`. El ordenador daría un error del tipo _you can't assign a string value ("Cuarenta y cinco mil") to a number variable (suscriptores_activos)_.

Comenzamos repasando los tipos de datos atómicos en Python (`str`, `int`, `float`, `bool`) aplicándolos a la gestión de datos de creadores de contenido en plataformas de _streaming_ en España.

### 2. Sugerencias de tipo o _Type hints_
Cuando un programa es pequeño, normalmente se pueden recordar los tipos de las variables. Pero a medida que los programas crecen es conveniente indicar el tipo de una variable para facilitar su lectura y aprovechar las funciones de los editores de código, que pueden indicar errores en función de estas sugerencias de tipo:

¿Es `alias` un str?
¿Qué tipo de valor numérico admite la media de la audiencia?
Las sugerencias de tipo nos permiten escribir esas expectativas directamente en nuestro código:


```python
# None
alias: str | None = "None"

# Float
espectadores_media: float = 78500

# Entero
suscriptores_activos: int = "45200"

# Booleano
emite_hoy: bool = 'True'
```


Las partes `emite_hoy: bool`, `alias: str | None` son sugerencias de tipo. Indican a los humanos y a los editores de código qué tipos de valores espera.

Las sugerencias de tipo no hacen que Python deje de ser Python . Sigue siendo un lenguaje de tipado dinámico y, como se puede ver en el ejemplo anterior, no rechazará automáticamente un valor incorrecto solo porque una sugerencia de tipo lo indique, pero probablemente vuestro editor os marcará el error para corregirlo.

Las sugerencias de tipo sirven para:

* Hacer que el código sea más fácil de leer.
* Ayudar a tu editor a autocompletar y avisar de los errores.
* Facilitar la detección de errores antes de ejecutar el código.

### 2.1. Listas y tuplas: Manipulación de secuencias

Las __listas__ son colecciones ordenadas y mutables de elementos. Pueden contener cualquier tipo de datos. Constituyen la estructura nativa fundamental para almacenar series de observaciones antes de trabajar con librerías especializadas como `numpy` o `pandas`.
A continuación, analizaremos una muestra de canciones con mayor número de reproducciones diarias en Spotify España.

```python
### 5. Trabajo con listas: mutabilidad, indexación y troceado (slicing)

"""
A veces, al crear listas manualmente, puede resultar difícil
leerlas si todos los elementos están en la misma línea de código.
Podemos declarar la lista usando varias líneas:
"""
canciones = [
    "BbY WOW",
    "La Graciosa",
    "De Lejitos - Remix",
    "Al Golpito"
]

# O en una sola línea:
reproducciones_millones = [0.535, 0.444, 0.361, 0.311]

# Acceso por índice e indexación negativa (último elemento)
print(f"Canción número 1 en España: {canciones[0]}")
print(f"Última canción de la lista: {canciones[-1]}\n")

# Tamaño de la lista
print(f"Lista con las {len(canciones)} canciones con más reproducciones:")
for cancion in canciones:
  print(cancion)

# Slicing: Obtener las dos primeras canciones
top_2 = canciones[:2]
print(f"\nTop 2 de la lista: {top_2}\n")

# Actualizar contenido de las listas
reproducciones_millones[0] = 0.602

# Añadir elementos a la lista
canciones.append("Vamo a vel")
reproducciones_millones.append(0.305)

# Eliminar (extraer) elementos de las listas
cancion_eliminada = canciones.pop()
# se le puede pasar el índice a eliminar
reproducciones_millones.pop(-1)
print(f"Canción eliminada: {cancion_eliminada}")
print(f"Valor de reproducciones último: {reproducciones_millones[-1]}\n")

# Operaciones de agregación nativas
total_reproducciones = sum(reproducciones_millones)
media_reproducciones = total_reproducciones / len(reproducciones_millones)
print(f"Total reproducciones acumuladas: {total_reproducciones:.2f} millones")
print(f"Promedio de reproducciones por tema: {media_reproducciones:.2f} millones")
```

Las __tuplas__ son estructuras de datos ordenadas pero inmutables —una vez creadas, sus elementos no pueden modificarse, añadirse ni eliminarse—. En _machine learning_ y desarrollo de software son fundamentales para almacenar registros fijos (como coordenadas o configuraciones) y para el retorno de múltiples valores en funciones.

```python
### 5. Trabajo con tuplas: inmutabilidad, indexación y desempaquetado

# Definición de una tupla:
# (Nombre del evento, Ciudad, Aforo máximo, Precio entrada €)
evento_velada = ("La Velada del Año IV", "Madrid", 80000, 55.0)

# 1. Acceso por índice e indexación negativa
print(f"Evento: {evento_velada[0]}")
print(f"Precio de la entrada básica: {evento_velada[-1]} €")

# 2. Troceado
localizacion = evento_velada[:2]
print(f"Datos de localización (Nombre y Ciudad): {localizacion}")

# 3. Demostración de inmutabilidad
# Intentar modificar un elemento de la tupla provocará un error de tipo (TypeError)
try:
  evento_velada[2] = 85000
except TypeError as error:
  print(
    f"\n[Demostración de inmutabilidad] No se puede modificar la tupla: {error}"
    )

# 4. Desempaquetado de tuplas
nombre, ciudad, aforo, precio = evento_velada
print(f"\n--- Ficha desempaquetada ---")
print(f"El evento «{nombre}» se celebrará en {ciudad}\ncon un aforo de {aforo:,} personas."
    )

# 5. Retorno múltiple de valores en funciones usando tuplas
def calcular_metricas_estadio(aforo_total, entradas_vendidas):
  """Calcula el porcentaje de ocupación y la recaudación estimada."""
  porcentaje_ocupacion = (entradas_vendidas / aforo_total) * 100
  recaudacion = entradas_vendidas * 55.0
  # Devuelve implícitamente una tupla: (porcentaje, recaudación)
  return porcentaje_ocupacion, recaudacion
```
